```python
def calculate(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        return a / b
    else:
        return "Invalid operation"


a = int(input())
b = int(input())
operation = input()

print(calculate(a, b, operation))
```

```python
def sum_numbers(*args):
    total = 0

    for num in args:
        total += num

    return total


print(sum_numbers(10, 20, 30, 40))
```

```python
def employee(**args):
    for key, value in args.items():
        print(key, ":", value)


employee(
    name="Reshma",
    ID=101,
    department="CSE",
    salary=30000
)
```

```python
def remove_duplicates(lst):
    result = []

    for item in lst:
        if item not in result:
            result.append(item)

    return result


lst = [10, 20, 10, 30, 20, 40, 30]

print(remove_duplicates(lst))
```
