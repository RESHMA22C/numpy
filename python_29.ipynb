```python
start = int(input())
end = int(input())

for num in range(start, end + 1):
    if num < 2:
        continue

    count = 0

    for i in range(2, num):
        if num % i == 0:
            count += 1

    if count == 0:
        print(num)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


n = int(input())

print(factorial(n))
```

```python
square = lambda x: x * x

n = int(input())

print(square(n))
```

```python
n = int(input())

a = list(map(int, input().split()))

largest = a[0]
second = a[0]

for i in range(1, n):
    if a[i] > largest:
        second = largest
        largest = a[i]
    elif a[i] > second and a[i] != largest:
        second = a[i]

print(second)
```

```python
s = input()

frequency = {}

for ch in s:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print(frequency)
```

```python
import math

radius = float(input())

area = math.pi * radius * radius

print(area)
```

```python
s = input()

reverse = ""

for i in range(len(s) - 1, -1, -1):
    reverse = reverse + s[i]

print(reverse)
```

```python
n = int(input())

a = list(map(int, input().split()))

result = []

for i in a:
    if i not in result:
        result.append(i)

print(result)
```

```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


n = int(input())

for i in range(n):
    print(fibonacci(i), end=" ")
```
